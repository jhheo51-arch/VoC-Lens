# VoC Lens 감정 모델 학습 (Colab GPU)

KOTE 4만 문장으로 한국어 감정 분류 모델을 파인튜닝합니다. 로컬 CPU 기본값은 작은 모델(koelectra-small)이고, 여기서는 댓글·리뷰 말투에 강한 **KcELECTRA-base**를 학습합니다.

1. 메뉴 **런타임 → 런타임 유형 변경 → T4 GPU**
2. 위에서부터 차례로 실행 (약 20~30분)
3. 마지막 셀이 만든 `emotion-kote-base.zip`을 내려받아 저장소의 `models/` 아래에 풉니다 (git에는 올리지 않음)
4. 로컬에서 `voclens eval kote data/external/kote/test.tsv --emotion-model models/emotion-kote-base` 로 비교
